# Der molekulare Hamilton-Operator

Als ersten Schritt in der Behandlung molekularer Systeme konstruieren wir den (nichtrelativistischen) molekularen Hamilton-Operator. Dieser besteht schematisch aus folgenden Beiträgen:

$$\require{color}
\fcolorbox{red}{#f2d5d3}{$ \hat{H}=\hat{T}_{n}+\hat{T}_{e}+\hat{V}_{ne}+\hat{V}_{ee}+\hat{V}_{nn}$} $$


Die einzelnen Terme stehen für:
- Kinetische Energie der Atomkerne:

$$ T_{n}=\sum_{I=1}^{M}\frac{|\mathbf{p}_{I}|^{2}}{2M_{I}}\rightarrow\hat{T}_{n}=\sum_{I=1}^{M}\frac{-\hbar^{2}}{2M_{I}}\nabla_{I}^{2}$$

- Kinetische Energie der Elektronen:

$$ T_{e}=\sum_{i=1}^{m}\frac{|\mathbf{p}_{i}|^{2}}{2m_{e}}\rightarrow\hat{T} _{e}=\sum_{i=1}^{m}\frac{-\hbar^{2}}{2m_{e}}\nabla_{i}^{2} $$

- Elektrostatische Elektron-Kern-Wechselwirkung (anziehend): 

$$ \hat{V}_{ne}=\frac{-e^{2}}{4\pi\epsilon_{0}}\sum_{I=1}^{M}\sum_{i=1}^{m}\frac{Z_{I}}{|\mathbf{R}_{I}-\mathbf{r}_{i}|} $$

-  Elektrostatische Elektron-Elektron-Wechselwirkung (abstoßend): 

$$\hat{V}_{ee}=\frac{e^{2}}{4\pi\epsilon_{0}}\sum_{i=1}^{m}\sum_{j>i}^{m}\frac{1}{|\mathbf{r}_{i}-\mathbf{r}_{j}|}$$

- Elektrostatische Kern-Kern-Wechselwirkung (abstoßend): 

$$\hat{V}_{nn}=\frac{e^{2}}{4\pi\epsilon_{0}}\sum_{I=1}^{M}\sum_{J>I}^{M}\frac{Z_I Z_J}{|\mathbf{R}_{I}-\mathbf{R}_{J}|} $$

Mit Hilfe dieser Operatoren konstruieren wir die nichtrelativistische molekulare Schrödingergleichung:

$$\left[-\sum_{I=1}^{M}\frac{\hbar^{2}}{2M_{I}}\nabla_{I}^{2}-\sum_{i=1}^{m}\frac{\hbar^{2}}{2m_{e}}\nabla_{i}^{2}+\frac{e^{2}}{4\pi\epsilon_{0}}\left(\sum_{I=1}^{M}\sum_{i=1}^{m}\frac{-Z_{N}}{r_{iI}}+\sum_{i=1}^{m}\sum_{j>i}^{m}\frac{1}{r_{ij}}+\sum_{I=1}^{M}\sum_{J>I}^{M}\frac{Z_{I}Z_{J}}{R_{IJ}}\right)\right]\Psi(\mathbf{r},\mathbf{R})	=	E\,\Psi(\mathbf{r},\mathbf{R})$$

Dabei haben wir die folgenden Abkürzungen definiert:

$$ r_{iI}=|\mathbf{R}_{I}-\mathbf{r}_{i}| $$

$$r_{ij}=|\mathbf{r}_{i}-\mathbf{r}_{j}|$$

$$ r_{IJ}=|\mathbf{R}_{I}-\mathbf{R}_{J}| $$

Die exakte <b>molekulare Wellenfunktion</b> $\Psi(\mathbf{r},\mathbf{R})$ ist im Allgemeinen eine komplizierte Funktion, die von $3(M+m)$ Variablen abhängt. $\mathbf{r}$ und $\mathbf{R}$ symbolisieren dabei die Gesamtheit der Ortsvektoren von Elektronen und Kernen.


Die Idee der Born-Oppenheimer-Näherung:

- Aus der Sicht der Elektronen sind die Kerne näherungsweise stationär (unbeweglich).

- Aus der Sicht der Kerne bewegen sich die Elektronen so schnell, dass sie vollkommen ''verschmiert'' sind

- Die Elektronen erzeugen ein gemitteltes effektives Potential in dem sich die Kerne bewegen.

Wir wollen im folgenden diese Vorstellungen, die intuitiv klar sind, mathematisch formulieren. 

Den exakten Hamilton-Operator des Moleküls können wir wie folgt zerlegen: 
 
$$ \hat{H}=\hat{T}_{n}+\hat{H}_{el}. $$

Dabei haben wir den **elektronischen Hamilton-Operator** definiert als:

$$ \hat{H}_{el}=\hat{T}_{e}+\hat{V}_{ne}+\hat{V}_{ee}+\hat{V}_{nn}. $$ 
    
Dieser Operator beschreibt physikalisch die Bewegung der Elektronen im **fixierten** elektrostatischen Feld der Kerne. Er enthält die Koordinaten der Kerne als multiplikative Faktoren, beinhaltet aber keine Ableitung nach den Kern-Koordinaten. Dies sollten wir uns gut merken.

Nun können wir im Prinzip die **elektronische Schrödingergleichung** lösen: 

$$\require{color}
\fcolorbox{red}{#f2d5d3}{$\hat{H}_{el}\psi_{el}^{(i)}(\mathbf{r};\mathbf{R})=E_{el}^{(i)}(\mathbf{R})\psi_{el}^{(i)}(\mathbf{r};\mathbf{R})\quad i=1,2,\cdots $}$$

Damit bekommen wir für **jede** Kernkonfiguration $\mathbf{R}$ einen Satz von elektronischen Zuständen. Die elektronische Energie $E_{el}^{(i)}(\mathbf{R})$ ist dabei abhängig von den Kernkoordinaten. Da wir für jede Kernkonfiguration einen anderen Satz von elektronischen Wellenfunktionen generieren, sind diese auch (parametrisch) abhängig von den Kernkoordinaten (daher die Bezeichnung $\psi^{(i)}_{el} (\mathbf{r};\mathbf{R})$ mit Semikolon statt Komma).

Die elektronischen Wellenfunktionen bilden für jede Kernkonfiguration einen orthonormierten Satz von Funktionen:

$$\require{colors}
\fcolorbox{red}{#f2d5d3}{$ \int d\mathbf{r}\,\psi_{el}^{(i)*}(\mathbf{r};\mathbf{R})\,\psi_{el}^{(j)}(\mathbf{r};\mathbf{R})=\delta_{ij}.$} $$

Man bemerke, dass dabei nur über elektronische Koordinaten integriert wird!

Da die elektronischen Wellenfunktionen eine vollständige orthonormierte Basis darstellen, kann die unbekannte exakte Wellenfunktion $\Psi (\mathbf{r},\mathbf{R})$ **exakt** in dieser Basis zerlegt werden:

$$\Psi(\mathbf{r},\mathbf{R})=\sum_{i}\chi_{i}(\mathbf{R})\,\psi_{el}^{(i)}(\mathbf{r};\mathbf{R}). $$

Die Entwicklungskoeffizienten $\chi_i(\mathbf{R})$ hängen nur von den Kern-Koordinaten ab. Man bezeichnet sie deswegen als **Kern-Wellenfunktionen**.

Da der elektronische Hamiltonoperator $\hat H_{el}$ generell unendlich viele Eigenfunktionen besitzt, enthält die Entwicklung im Allgemeinen unendlich viele Terme.

<div class="alert alert-block alert-warning">
Die Born-Oppenheimer-Näherung besteht nun darin, die unendliche Entwicklung durch <b> einen einzigen Term </b> der Form

$$\Psi(\mathbf{r},\mathbf{R})=\sum_{i}\chi_{i}(\mathbf{R})\psi_{el}^{(i)}(\mathbf{r};\mathbf{R})\rightarrow\Psi_{BO}(\mathbf{r},\mathbf{R})=\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})  $$
    zu ersetzen.
</div>

Die molekulare Wellenfunktion wird also als Produkt einer elektronischen und einer Kern-Wellenfunktion dargestellt. Die elektronische Wellenfunktion bekommt man als Lösung der elektronischen Schrödingergleichung. Aber wo kommt die Kern-Wellenfunktion her?

Wir betrachten nun die Born-Oppenheimer-Wellenfunktion als einen Ansatz und bestimmen $\chi(\mathbf{R})$ variationell. Unter der Annahme, dass der Ansatz normiert ist, kann man die Energie wie folgt bestimmen:

$$ E[\chi]=\int d\mathbf{R}\int d\mathbf{r}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]^{*}\hat{H}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right] $$

Um unsere Intuition mathematisch umzusetzen, die uns sagt, dass sich die Kerne in einem **effektiven Potential** bewegen, das von den Elektronen erzeugt wird, versuchen wir, die Energie in der folgenden Form auszudrücken, in der die elektronischen Koordinaten herausintegriert sind:

$$ E[\chi]=\int d\mathbf{R}\,\chi^{*}(\mathbf{R})\hat{H}_{nuc}\chi(\mathbf{R}). $$

$H_{nuc}$ ist der effektive Kern-Hamiltonoperator, dessen genaue Form wir nun bestimmen müssen!

Wenn wir den Energieausdruck ausschreiben, bekommen wir:

$$E[\chi]=\int d\mathbf{R}\int d\mathbf{r}\,\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]^{*}\left(\hat{T}_{n}+\hat{H}_{el}\right)\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]$$


$$E[\chi]=\int d\mathbf{R}\int d\mathbf{r}\,\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]^{*}\hat{T}_{n}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]+\int d\mathbf{R}\int d\mathbf{r}\,\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]^{*}\hat{H}_{el}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]$$


Nun können wir ausnutzen, dass die elektronische Wellenfunktion eine Eigenfunktion des elektronischen Hamiltonoperators ist:

$$\hat{H}_{el}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]=\chi(\mathbf{R})\hat{H}_{el}\psi_{el}(\mathbf{r};\mathbf{R})=\chi(\mathbf{R})E_{el}(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})$$

Dies, zusammen mit der Normierung der elektronischen Wellenfunktion

$$\int d\mathbf{r}\,\psi_{el}^{*}(\mathbf{r};\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})=1$$

bringt schon die zweite Hälfte des Energieausdrucks auf die gewünschte Form, in der keine elektronischen Koordinaten mehr vorkommen:

$$\require{color}E[\chi]=\int d\mathbf{R}\int d\mathbf{r}\,\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]^{*}\hat{T}_{n}\left[\chi(R)\psi_{el}(\mathbf{r};\mathbf{R})\right]+\fcolorbox{blue}{red!50}{$\int d\mathbf{R}\,\chi^{*}(\mathbf{R})E_{el}(\mathbf{R})\chi(\mathbf{R})$}$$

Als nächstes widmen wir uns dem ersten, komplizierteren Term.

Die Anwendung des Operators der kinetischen Energie der Kerne auf das Born-Oppenheimer-Produkt führt zu:

$$\hat{T}_{n}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]=\sum_{I}\left(-\frac{\hbar^{2}}{2M_{I}}\right)\nabla_{I}^{2}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]$$

Die Wirkung des Nabla-Operators können wir explizit durch zweifaches Ableiten ausrechnen:

$$\nabla_{I}^{2}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]=\left(\nabla_{I}^{2}\chi\right)\psi_{el}+2\nabla_{I}\chi\cdot\nabla_{I}\psi_{el}+\chi\left(\nabla_{I}^{2}\psi_{el}\right)$$

und bekommen damit:

$$\hat{T}_{n}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]=\sum_{I}\left(-\frac{\hbar^{2}}{2M_{I}}\right)\left[\left(\nabla_{I}^{2}\chi\right)\psi_{el}+2\nabla_{I}\chi\cdot\nabla_{I}\psi_{el}+\chi\left(\nabla_{I}^{2}\psi_{el}\right)\right]$$

Einsetzen in den ersten Term des Energieausdrucks liefert:


$$\int d\mathbf{R}\int d\mathbf{r}\,\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]^{*}\hat{T}_{n}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]=\int d\mathbf{R}\chi^{*}(\mathbf{R})\left[\sum_{I}\left(-\frac{\hbar^{2}}{2M_{I}}\right)\left(\nabla_{I}^{2}\chi\right)\int d\mathbf{r}\,\psi_{el}^{*}\psi_{el}-\sum_{I}\left(\frac{\hbar^{2}}{M_{I}}\right)\nabla_{I}\chi\cdot\int d\mathbf{r}\,\psi_{el}^{*}\nabla_{I}\psi_{el}+\chi\sum_{I}\left(-\frac{\hbar^{2}}{2M_{I}}\right)\int d\mathbf{r}\,\psi_{el}^{*}\left(\nabla_{I}^{2}\psi_{el}\right)\right]$$

Dies können wir vereinfachen, indem wir die folgenden zwei Größen definieren:

$$\mathbf{A}_{I}(\mathbf{R})\equiv i\hbar\int d\mathbf{r}\,\psi_{el}^{*}\nabla_{I}\psi_{el}$$

$$D_{I}(\mathbf{R})\equiv\int d\mathbf{r}\,\psi_{el}^{*}\nabla_{I}^{2}\psi_{el}$$

Die erste Größe $\mathbf{A}_{I}$ ist ein Vektor und wird in der Literatur als "Berry's connection" bezeichnet. Die zweite Größe ist ein Skalar. Mit diesen zwei Definitionen kann man den kinetischen Term wie folgt umformen:

$$\int d\mathbf{R}\int d\mathbf{r}\,\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]^{*}\hat{T}_{n}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]=\int d\mathbf{R}\chi^{*}(\mathbf{R})\left[\sum_{I}\left(-\frac{\hbar^{2}}{2M_{I}}\right)\left(\nabla_{I}^{2}\chi\right)-\sum_{I}\left(\frac{-i\hbar}{M_{I}}\right)\nabla_{I}\chi\cdot\mathbf{A}_{I}(\mathbf{R})+\sum_{I}\left(-\frac{\hbar^{2}}{2M_{I}}\right)D_{I}(\mathbf{R})\chi\right] $$

und bekommen damit:

$$\int d\mathbf{R}\int d\mathbf{r}\,\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]^{*}\hat{T}_{n}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]=\int d\mathbf{R}\,\chi^{*}(\mathbf{R})\left[\sum_{I}\left(\frac{\mathbf{P}_{I}^{2}\chi}{2M_{I}}\right)-\sum_{I}\left(\frac{\mathbf{P}_{i}\chi}{M_{I}}\right)\cdot\mathbf{A}_{I}(\mathbf{R})+\sum_{I}\left(-\frac{\hbar^{2}}{2M_{I}}\right)D_{I}(\mathbf{R})\chi\right]$$

Um dies weiter zu vereinfachen, betrachten wir das folgende Integral:

$$ \int d\mathbf{r}\,\nabla_{I}\cdot\left(\psi_{el}^{*}\nabla_{I}\psi_{el}\right) $$

Wir erinnern uns an die Definition des Gradienten einer skalaren Funktion ($\varphi$) und der Divergenz einer Vektorfunktion ($\mathbf{f}$):

$$\nabla\varphi=\frac{\partial\varphi}{\partial x}\mathbf{e}_{x}+\frac{\partial\varphi}{\partial y}\mathbf{e}_{y}+\frac{\partial\varphi}{\partial z}\mathbf{e}_{z} $$

$$ \nabla\cdot\mathbf{f}=\frac{\partial f_{x}}{\partial x}+\frac{\partial f_{y}}{\partial y}+\frac{\partial f_{z}}{\partial z}$$

Hier bezeichnen $\mathbf{e}_{x}$, $\mathbf{e}_{y}$ und $\mathbf{e}_{z}$ Einheitsvektoren in die $x$-, $y$- und $z$-Richtung.

Die Anwendung dieser Ausdrücke liefert nun:

$$\nabla_{I}\psi_{el}=\frac{\partial\psi_{el}}{\partial x_{I}}\mathbf{e}_{x,I}+\frac{\partial\psi_{el}}{\partial y_{I}}\mathbf{e}_{y,I}+\frac{\partial\psi_{el}}{\partial z_{I}}\mathbf{e}_{z,I}$$

$$\psi_{el}^{*}\nabla_{I}\psi_{el}=\psi_{el}^{*}\frac{\partial\psi_{el}}{\partial x_{I}}\mathbf{e}_{x,I}+\psi_{el}^{*}\frac{\partial\psi_{el}}{\partial y_{I}}\mathbf{e}_{y,I}+\psi_{el}^{*}\frac{\partial\psi_{el}}{\partial z_{I}}\mathbf{e}_{z,I}$$

Daraus folgt, 

$$ \nabla_{I}\cdot\left(\psi_{el}^{*}\nabla_{I}\psi_{el}\right)=\frac{\partial}{\partial x_{I}}\left[\psi_{el}^{*}\frac{\partial\psi_{el}}{\partial x_{I}}\right]+\frac{\partial}{\partial y_{I}}\left[\psi_{el}^{*}\frac{\partial\psi_{el}}{\partial y_{I}}\right]+\frac{\partial}{\partial z_{I}}\left[\psi_{el}^{*}\frac{\partial\psi_{el}}{\partial z_{I}}\right]$$

Dies kann man kürzer als

\begin{eqnarray}
\nabla_{I}\cdot\left(\psi_{el}^{*}\nabla_{I}\psi_{el}\right)&=&\psi_{el}^{*}\nabla_{I}^{2}\psi_{el}+\nabla_{I}\psi_{el}^{*}\cdot\nabla_{I}\psi_{el}\\
&=&\psi_{el}^{*}\nabla_{I}^{2}\psi_{el}+|\nabla_{I}\psi_{el}|^{2}
\end{eqnarray}

schreiben. 
    

Durch Integration über die elektronischen Koordinaten erhalten wir: 

$$\int d\mathbf{r}\,\nabla_{I}\cdot\left(\psi_{el}^{*}\nabla_{I}\psi_{el}\right)=\int d\mathbf{r}\,\psi_{el}^{*}\nabla_{I}^{2}\psi_{el}+\int d\mathbf{r}\,|\nabla_{I}\psi_{el}|^{2}$$

Nun können wir in diesen Ausdruck unsere Definitionen von $D_I$ und $\mathbf{A}_{I}$ einführen:

\begin{eqnarray}
D_{I}(\mathbf{R})&=&\int d\mathbf{r}\,\psi_{el}^{*}\nabla_{I}^{2}\psi_{el}\\
&=&\nabla_{I}\cdot\int d\mathbf{r}\,\psi_{el}^{*}\nabla_{I}\psi_{el}-\int d\mathbf{r}\,|\nabla_{I}\psi_{el}|^{2}\\
&=&\frac{1}{i\hbar}\nabla_{I}\cdot\mathbf{A}_{I}-\int d\mathbf{r}\,|\nabla_{I}\psi_{el}|^{2}
\end{eqnarray}



Der kinetische Term des Erwartungswertes lässt sich damit in die folgende Form überführen:

$$\int d\mathbf{R}\int d\mathbf{r}\,\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]^{*}\hat{T}_{n}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]=\int d\mathbf{R}\chi^{*}(\mathbf{R})\left[\sum_{I}\left(\frac{\mathbf{P}_{I}^{2}\chi}{2M_{I}}\right)-\sum_{I}\left(\frac{\mathbf{P}_{i}\chi}{M_{I}}\right)\cdot\mathbf{A}_{I}(\mathbf{R})+\sum_{I}\left(-\frac{\hbar^{2}}{2M_{I}}\right)\left(\frac{1}{i\hbar}\nabla_{I}\cdot\mathbf{A}_{I}-\int d\mathbf{r}\,|\nabla_{I}\psi_{el}|^{2}\right)\chi\right]$$

$$\int d\mathbf{R}\int d\mathbf{r}\,\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]^{*}\hat{T}_{n}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]=\int d\mathbf{R}\chi^{*}(\mathbf{R})\left[\sum_{I}\left(\frac{\mathbf{P}_{I}^{2}\chi}{2M_{I}}\right)-\sum_{I}\left(\frac{\mathbf{P}_{i}\chi}{M_{I}}\right)\cdot\mathbf{A}_{I}(\mathbf{R})+\sum_{I}\left(-\frac{\mathbf{P}_{I}\cdot\mathbf{A}_{I}}{2M_{I}}\right)\chi+\sum_{I}\frac{\hbar^{2}}{2M_{I}}\int d\mathbf{r}\,|\nabla_{I}\psi_{el}|^{2}\chi\right]$$

Die ersten drei Terme lassen sich zu einem quadratischen Ausdruck erweitern.


$$\int d\mathbf{R}\int d\mathbf{r}\,\left[\chi(R)\psi_{el}(\mathbf{r};\mathbf{R})\right]^{*}\hat{T}_{n}\left[\chi(\mathbf{R})\psi_{el}(\mathbf{r};\mathbf{R})\right]=\int d\mathbf{R}\,\chi^{*}(\mathbf{R})\left[\sum_{I}\frac{(\mathbf{P}_{I}-\mathbf{A}_{I})^{2}}{2M_{I}}+\sum_{I}\frac{\hbar^{2}}{2M_{I}}\left(\int d\mathbf{r}\,|\nabla_{I}\psi_{el}|^{2}\right)-\sum_{I}\frac{|\mathbf{A}_{I}|^{2}}{2M_{I}}\right]\chi$$

Damit ist die Energie des BO-Ansatzes:

$$ E[\chi ]=\int d\mathbf{R}\chi^{*}(\mathbf{R})\left[\sum_{I}\frac{(\mathbf{P}_{I}-\mathbf{A}_{I})^{2}}{2M_{I}}+E_{el}(\mathbf{R})+\sum_{I}\frac{\hbar^{2}}{2M_{I}}\left(\int d\mathbf{r}\,|\nabla_{I}\psi_{el}|^{2}\right)-\sum_{I}\frac{|\mathbf{A}_{I}|^{2}}{2M_{I}}\right]\chi $$

Wir sehen, dass darin die elektronischen Koordinaten nicht mehr vorkommen. Wir haben sie "wegintegriert" und haben einen **effektiven** Hamilton-Operator für die Kernbewegung bekommen:

$$
\hat{H}_{nuc}=\sum_{I}\frac{(\mathbf{P}_{I}-\mathbf{A}_{I})^{2}}{2M_{I}}+E_{el}(\mathbf{R})+\sum_{I}\frac{\hbar^{2}}{2M_{I}}\left(\int d\mathbf{r}\,|\nabla_{I}\psi_{el}|^{2}\right)-\sum_{I}\frac{|\mathbf{A}_{I}|^{2}}{2M_{I}}
$$

mit dem man das Energiefunktional wie folgt schreiben kann:

$$
E[\chi]=\int d\mathbf{R}\chi^{*}(\mathbf{R})\hat{H}_{nuc}\chi(\mathbf{R})
$$

Wie müssen die Kern-Wellenfunktion nun so wählen, dass der Erwartungswert des effektiven Kern-Hamiltonoperators minimal wird. Aus dem Variationsprinzip folgt, dass der minimale Wert erreicht wird, wenn die Kern-Wellenfunktion $\chi $ eine **Eigenfunktion** des effektiven Kern-Hamiltonoperators ist:

$$ \hat{H}_{nuc}\chi(\mathbf{R})=E\chi(\mathbf{R}) $$

$$\sum_{I}\frac{(\mathbf{P}_{I}-\mathbf{A}_{I})^{2}}{2M_{I}}\chi+E_{el}(\mathbf{R})\chi+\sum_{I}\frac{\hbar^{2}}{2M_{I}}\left(\int d\mathbf{r}|\nabla_{I}\psi_{el}|^{2}\right)\chi-\sum_{I}\frac{|\mathbf{A}_{I}|^{2}}{2M_{I}}\chi=E\chi $$

Diese Schrödinger-Gleichung bezeichnet man als "Kern-Schrödingergleichung". Die entsprechenden Eigenwerte stellen die gequantelten Energien des Moleküls dar!

Diese Schrödingergleichung ist immer noch sehr kompliziert, so dass man üblicherweise noch weitere Näherungen einführt. Wenn die elektronische Wellenfunktion reell ist, kann man zeigen, dass die "Berry's connection" verschwindet.

$$ \psi_{el}\;\mathrm{reell}\rightarrow\mathbf{A}_{I}=0 $$

Wir erwarten intuitiv, dass sich die elektronische Wellenfunktion ''langsam'' mit den Kernkoordinaten verändert. In diesem Fall kann man auch $|\nabla_{I}\psi_{el}|^{2} $ vernachlässigen. Der effektive Kern-Hamiltonoperator vereinfacht sich damit weiter zu:

$$ \hat{H}_{nuc}=\sum_{I}\frac{\mathbf{P}_{I}{}^{2}}{2M_{I}}+E_{el}(\mathbf{R})=\hat{T}_{n}+E_{el}(\mathbf{R}) $$

Dieser Operator besteht nun nur aus der kinetischen Energie der Kerne, die sich in einem effektiven Potential $E_{el}$ bewegen, das durch die Elektronen erzeugt wird.
