# Das Helium-Atom

Nachdem wir das Variationsprinzip als Werkzeug zur näherungsweisen Lösung quantenmechanischer Probleme kennengelernt haben, wenden wir uns nun dem Heliumatom zu. Obwohl es nur zwei Elektronen besitzt, ist es aufgrund der gegenseitigen Elektronenabstoßung deutlich schwieriger zu behandeln als das Wasserstoffatom. Bereits an diesem einfachen Beispiel begegnen uns zentrale Herausforderungen der Quantenchemie.

Das Heliumatom eignet sich besonders gut, um grundlegende Konzepte und Näherungsverfahren zu entwickeln, da sich viele Rechenschritte noch „von Hand“ nachvollziehen lassen. Die dabei gewonnenen Einsichten und Ansätze werden wir anschließend auf größere Mehrelektronenatome und molekulare Systeme übertragen und verallgemeinern.

Im [vorangegangenen Kapitel](quasiteilchen_correlation.ipynb) haben wir am Beispiel gekoppelter Oszillatoren gesehen, wie kollektive Koordinaten eine Separation ermöglichen und wie die ursprünglichen Auslenkungen dennoch korreliert sein können. Nun prüfen wir, was sich davon auf zwei Elektronen im Feld eines Atomkerns übertragen lässt, und entwickeln anschließend schrittweise Näherungen für ihre Grundzustandsenergie.


## Der atomare Hamiltonoperator


Das Heliumatom besteht aus einem Kern mit Ladungszahl $Z = 2$ und zwei Elektronen.  
Wir behandeln den Kern als fest und vernachlässigen relativistische sowie spinabhängige Beiträge. Der elektronische Hamiltonoperator lautet in atomaren Einheiten  
(a.u., also $\hbar = m_e = e = 4\pi\varepsilon_0 = 1$):

$$
H = \sum_{i=1}^2 \left(-\tfrac12 \nabla_i^2 - \frac{Z}{r_i}\right) + \frac{1}{r_{12}},
\qquad r_i = |\mathbf r_i|,\quad r_{12}=|\mathbf r_1-\mathbf r_2|.
$$

## Schwerpunkt- und Relativkoordinaten

Kann eine geeignete Koordinatenwahl die Kopplung wie bei den harmonischen Oszillatoren beseitigen? Um dies zu untersuchen, betrachten wir zunächst zwei Elektronen ohne äußeres Potential, die sich nur gegenseitig abstoßen. Der Hamiltonoperator lautet dann

$$
\hat H=-\frac12(\nabla_1^2+\nabla_2^2)
+\frac{1}{r_{12}}.
$$

Wir führen den Schwerpunkt $\mathbf R$ und den Relativvektor $\mathbf r$ ein. Da beide Elektronen die Masse eins besitzen, sind die Gesamtmasse $M$ und die reduzierte Masse $\mu$ gegeben durch

$$
\mathbf R=\frac{\mathbf r_1+\mathbf r_2}{2},
\qquad\mathbf r=\mathbf r_1-\mathbf r_2,
\qquad M=2,\quad\mu=\frac12.
$$

Damit erhalten wir

$$
\hat H=-\frac14\nabla_R^2
+\left(-\nabla_r^2+\frac{1}{r}\right).
$$

**Der Hamiltonoperator separiert in Schwerpunkt- und Relativbewegung.** Die Coulomb-Abstoßung hängt nur vom Elektronenabstand $r=r_{12}$ ab und lässt die Schwerpunktbewegung frei. Das rein abstoßende Coulomb-Potential besitzt keine gebundenen relativen Zustände; es beschreibt Streuzustände. Wir erhalten daher kein gebundenes Elektronenpaar.


### Der Atomkern koppelt Schwerpunkt und innere Bewegung

Nun setzen wir den Kern der Ladung $+Z$ wieder in den Ursprung; für Helium ist $Z=2$. Damit erhalten wir den oben eingeführten atomaren Hamiltonoperator, den wir im Folgenden auch mit $\hat H_{\mathrm{He}}$ bezeichnen.

In den ursprünglichen Elektronenkoordinaten sind die beiden Kernanziehungen einfach; dafür koppelt $1/r_{12}$ die Elektronen. Versuchen wir erneut die Schwerpunkt- und Relativkoordinaten, so folgt aus

$$
\mathbf r_1=\mathbf R+\frac{\mathbf r}{2},
\qquad
\mathbf r_2=\mathbf R-\frac{\mathbf r}{2}
$$

die Darstellung

$$
\boxed{
\begin{aligned}
\hat H_{\mathrm{He}}
={}&-\frac14\nabla_R^2
-\nabla_r^2+\frac{1}{r}\\
&-\frac{Z}{|\mathbf R+\mathbf r/2|}
-\frac{Z}{|\mathbf R-\mathbf r/2|}.
\end{aligned}
}
$$

Die Elektronenabstoßung hängt jetzt nur von $r$ ab. Dafür enthalten die **Kernanziehungen beide Koordinaten gleichzeitig**. Denn mit dem Winkel $\theta$ zwischen $\mathbf R$ und $\mathbf r$ gilt

$$
\left|\mathbf R\pm\frac{\mathbf r}{2}\right|^2
=R^2+\frac{r^2}{4}\pm\mathbf R\cdot\mathbf r
=R^2+\frac{r^2}{4}\pm Rr\cos\theta.
$$

Ändern wir bei festem $R$ und $r$ die Orientierung des Relativvektors, ändern sich die Kernabstände. Eine Änderung der inneren Anordnung beeinflusst somit die Wechselwirkung des Paares mit dem Kern. Das Kernpotential lässt sich nicht als Summe $V_R(\mathbf R)+V_r(\mathbf r)$ schreiben: Wegen der Rotationssymmetrie könnten getrennte Beiträge nur von den Beträgen $R$ beziehungsweise $r$ abhängen. Eine Abhängigkeit vom relativen Winkel ist damit nicht vereinbar.

Hier bezeichnet $\mathbf R$ den **Schwerpunkt der beiden Elektronen relativ zum festen Kern**. Bewegt sich auch der Kern, lässt sich die freie Translation des gesamten Atoms weiterhin abtrennen. Das verbleibende innere Dreiteilchenproblem bleibt gekoppelt.


### Vergleich mit einer harmonischen Falle

Um die Ursache dieser Kopplung genauer zu verstehen, ersetzen wir die Kernanziehung vorübergehend durch dasselbe isotrope harmonische Potential für beide Elektronen:

$$
V_{\mathrm{Falle}}=\frac12\omega^2(r_1^2+r_2^2).
$$

Nun gilt

$$
r_1^2+r_2^2
=\left|\mathbf R+\frac{\mathbf r}{2}\right|^2
+\left|\mathbf R-\frac{\mathbf r}{2}\right|^2
=2R^2+\frac{r^2}{2}.
$$

Die Skalarprodukte heben sich auf. Trotz der weiterhin vorhandenen Coulomb-Abstoßung erhalten wir

$$
\begin{aligned}
\hat H_{\mathrm{Falle}}
={}&\underbrace{\left(-\frac14\nabla_R^2
+\omega^2R^2\right)}_{\hat H_R}\\
&+\underbrace{\left(-\nabla_r^2
+\frac14\omega^2r^2+\frac{1}{r}\right)}_{\hat H_r}.
\end{aligned}
$$

Schwerpunkt- und Relativbewegung sind **exakt getrennt**. Das relative Problem enthält aber weiterhin die Elektronenabstoßung und ist kein einfacher harmonischer Oszillator. Auch eine Produktform in $\mathbf R,\mathbf r$ macht die Elektronenorte nicht unabhängig.

Die folgende Abbildung vergleicht die Winkelabhängigkeit beider äußeren Potentiale bei festen Beträgen $R=r=2$ a.u. Links sind die Koordinaten für $\theta=60^\circ$ eingezeichnet. Rechts ist jeweils die Änderung gegenüber $\theta=90^\circ$ dargestellt:

$$
\Delta V(u)=V(u)-V(0),\qquad u=\cos\theta.
$$

Die Energieänderung $\Delta V$ ist in atomaren Einheiten angegeben, also in Hartree. Beim Kernpotential verändert eine Drehung des Relativvektors die Energie; bei der harmonischen Falle bleibt sie gleich. Die Elektronenabstoßung $1/r$ ist bei festem $r$ in beiden Fällen winkelunabhängig.


```python
%matplotlib inline
def plot_nuclear_coupling():
    import numpy as np
    import matplotlib.pyplot as plt
    from matplotlib.patches import Arc

    R_length, r_length, Z = 2.0, 2.0, 2.0  # Atomare Einheiten: Längen in a_0, Energien in Hartree
    theta = np.pi / 3
    R_vector = np.array([R_length, 0.0])
    r_vector = r_length * np.array([np.cos(theta), np.sin(theta)])
    electron_1 = R_vector + r_vector / 2
    electron_2 = R_vector - r_vector / 2
    u = np.linspace(-1, 1, 501)
    nuclear_energy = -Z / np.sqrt(R_length**2 + r_length**2 / 4 + R_length * r_length * u)
    nuclear_energy -= Z / np.sqrt(R_length**2 + r_length**2 / 4 - R_length * r_length * u)
    reference = -2 * Z / np.sqrt(R_length**2 + r_length**2 / 4)
    style = {"font.family": "DejaVu Serif", "font.size": 11,
             "mathtext.fontset": "dejavuserif", "figure.dpi": 150}

    with plt.rc_context(style):
        fig, (geometry, energy) = plt.subplots(
            1, 2, figsize=(10, 4.1), layout="constrained",
            gridspec_kw={"width_ratios": [1, 1.25]})
        for point in [electron_1, electron_2]:
            geometry.plot([0, point[0]], [0, point[1]], color="0.65", lw=1)
        geometry.annotate("", xy=R_vector, xytext=(0, 0),
                          arrowprops={"arrowstyle": "->", "color": "black", "lw": 1.8})
        geometry.annotate("", xy=electron_1, xytext=electron_2,
                          arrowprops={"arrowstyle": "->", "color": "#3366aa", "lw": 2})
        geometry.plot(0, 0, "o", color="black", ms=9)
        geometry.plot(*R_vector, "o", color="white", mec="black", ms=6)
        geometry.plot(*electron_1, "o", color="#b65447", ms=9)
        geometry.plot(*electron_2, "o", color="#3366aa", ms=9)
        geometry.plot([2, 2.7], [0, 0], ls=":", color="gray", lw=1)
        geometry.add_patch(Arc(R_vector, 0.85, 0.85, theta1=0, theta2=60,
                               color="black", lw=1))
        geometry.text(-0.05, -0.25, "Kern", ha="center")
        geometry.text(0.95, -0.22, r"$\mathbf{R}$", ha="center")
        geometry.text(2.03, -0.28, "Paarzentrum", ha="left", fontsize=10)
        geometry.text(1.93, 0.56, r"$\mathbf{r}$", color="#3366aa")
        geometry.text(2.49, 0.20, r"$\theta$")
        geometry.text(electron_1[0] + 0.1, electron_1[1] + 0.06,
                      r"$e_1$", color="#b65447")
        geometry.text(electron_2[0] - 0.28, electron_2[1] - 0.14,
                      r"$e_2$", color="#3366aa")
        geometry.set(xlim=(-0.35, 3.2), ylim=(-1.35, 1.35), aspect="equal")
        geometry.set_title(r"$\mathbf{r}_{1,2}=\mathbf{R}\pm\mathbf{r}/2$", pad=18)
        geometry.axis("off")

        energy.plot(u, nuclear_energy - reference, color="#b65447", lw=2.3,
                    label="Coulomb-Kernpotential")
        energy.axhline(0, color="#3366aa", ls="--", lw=2,
                       label="Harmonische Falle")
        energy.set(xlim=(-1, 1), ylim=(-0.98, 0.12),
                   xlabel=r"$u=\cos\theta$",
                   ylabel=r"$V(u)-V(0)$ (a.u.)",
                   title=r"Feste Abstände: $R=r=2$ a.u.")
        energy.set_xticks([-1, 0, 1], ["−1\nantiparallel", "0\nsenkrecht", "1\nparallel"])
        energy.set_yticks([-0.8, -0.4, 0])
        energy.spines[["top", "right"]].set_visible(False)
        energy.grid(axis="y", color="0.87", lw=0.8)
        energy.set_axisbelow(True)
        energy.legend(loc="lower center", frameon=False, fontsize=10)
        plt.show()


plot_nuclear_coupling()
```

### Was ist an den Normalkoordinaten besonders?

Für ein quadratisches Potential ist die Matrix der zweiten Ableitungen, die **Hesse-Matrix**, konstant. Beim gekoppelten Oszillator ist sie

$$
\mathbf K=
\begin{pmatrix}
m\omega_0^2&\lambda\\
\lambda&m\omega_0^2
\end{pmatrix}.
$$

Eine feste orthogonale Transformation diagonalisiert diese Matrix überall im Konfigurationsraum. Bei den hier gleichen Massen bleibt dabei auch die kinetische Energie diagonal. Für unterschiedliche Massen verwendet man zuvor massengewichtete Koordinaten.

Beim Coulomb-Potential hängt die Krümmung dagegen vom Ort ab. Eine lokale quadratische Näherung und ihre Normalmoden liefern daher keine globale exakte Entkopplung. **Nichtquadratisch bedeutet jedoch nicht grundsätzlich unseparierbar:** Das freie Coulomb-Paar hat gerade gezeigt, dass eine andere Struktur, hier die Translationsinvarianz, die Separation ermöglichen kann.

Auch nichtlineare Koordinaten können ein Problem vereinfachen. Für den rotationsinvarianten räumlichen Heliumgrundzustand sind die drei Abstände $r_1,r_2,r_{12}$ besonders zweckmäßig. In diesen Variablen ist das Potential einfach, aber die kinetische Energie enthält gekoppelte Ableitungen. Zudem sind die Abstände durch die Dreiecksungleichung

$$
|r_1-r_2|\le r_{12}\le r_1+r_2
$$

eingeschränkt. Die Wahl passender Koordinaten beseitigt somit nicht automatisch die Kopplung des gesamten Problems.


## Näherung ohne Elektron–Elektron-Abstoßung

Die Schwerpunkt- und Relativkoordinaten zerlegen das elektronische Heliumproblem also nicht wie die Normalkoordinaten der gekoppelten Oszillatoren in unabhängige Teilprobleme. Wir wenden uns deshalb einer näherungsweisen Behandlung zu.

Als Ausgangspunkt für unsere Behandlung des He-Atoms lassen wir im ersten Schritt die Elektron–Elektron-Abstoßung vollständig **weg** und erhalten damit den folgenden vereinfachten Hamiltonian:

$$
H_0 = \sum_{i=1}^2 \left(-\tfrac12 \nabla_i^2 - \frac{Z}{r_i}\right).
$$

In diesem Hamiltonian sind beide Elektronen vollständig entkoppelt, und der Operator kann als Summe zweier Terme geschrieben werden, die jeweils nur von den Koordinaten des einen oder des anderen Elektrons abhängen.  
Solche Terme bezeichnet man in der Quantenchemie als **Ein-Elektron-Terme**. Es gilt also

$$
H_0 = h(1) + h(2),
$$

mit dem bekannten wasserstoffähnlichen Hamiltonian

$$
h = -\tfrac12 \nabla^2 - \frac{Z}{r}.
$$

Die Eigenfunktionen und Eigenwerte dieses Ein-Elektron-Hamiltonoperators kennen wir bereits vom Wasserstoffatom. Die folgende Übersicht zeigt ausgewählte **wasserstoffähnliche Orbitale** und ihre normierten analytischen Ausdrücke für eine beliebige Kernladungszahl $Z$:

Das 2s-Schnittbild zeigt den radialen Knoten bei r = 2/Z; Blau und Koralle kennzeichnen positives und negatives Vorzeichen der Wellenfunktion.](figures/hydrogenic_orbitals/hydrogenic_orbitals_cartoon.png)

*Alle Ausdrücke gelten in atomaren Einheiten; $E_n=-Z^2/(2n^2)$ ist in Hartree angegeben. Die reellen p- und d-Orbitale lassen sich aus den Kugelflächenfunktionen $Y_{lm}$ bilden. Für den räumlichen Grundzustand unseres vereinfachten Heliumatoms besetzen beide Elektronen das 1s-Orbital mit $Z=2$.*

Lässt man auch bei einem allgemeinen N-Elektronen-Atom die Elektron–Elektron-Wechselwirkung fort, so ergibt sich der ungekoppelte Hamiltonoperator zu

$$
H_0 = \sum_{i=1}^{N} \left(-\tfrac12 \nabla_i^2 - \frac{Z}{r_i}\right)
    = \sum_{i=1}^{N} h(i),
$$

wobei jeder Summand $h(i)$ den Ein-Elektron-Hamiltonian für das $i$-te Elektron beschreibt:

$$
h(i) = -\tfrac12 \nabla_i^2 - \frac{Z}{r_i}.
$$

## Separationsansatz (Hartree-Produkt)

Um den Grundzustand des Heliumatoms näherungsweise zu beschreiben, wählen wir für die **räumliche Wellenfunktion** einen **Produktansatz**, den man auch als **Hartree-Produkt** bezeichnet:

$$
\Psi(\mathbf r_1,\mathbf r_2) = \phi(\mathbf r_1)\,\phi(\mathbf r_2).
$$

Dabei ist $\phi$ ein normiertes, 1s-ähnliches Atomorbital mit Zentrum am Kern. Beide Elektronen werden somit durch dasselbe räumliche Orbital beschrieben. Ist $\phi$ normiert, so ist auch das Hartree-Produkt normiert. Wir ignorieren momentan die Tatsache, dass Elektronen auch einen Spin besitzen. Das Ergebniss das wir erhalten bleibt in diesem Fall auch dann korrekt, wenn wir den Spin später berücksichtigen.

**Bemerkung:** Warum verwenden wir ein Produkt und nicht einfach eine Summe der Form

$$
\Psi(\mathbf r_1,\mathbf r_2) = \phi(\mathbf r_1) + \phi(\mathbf r_2)?
$$

Da das Orbital im Unendlichen verschwindet, gilt für diese Summe

$$
\lim_{r_2 \to \infty} \Psi(\mathbf r_1,\mathbf r_2)
= \phi(\mathbf r_1).
$$

Die Wellenfunktion verschwindet also im Allgemeinen nicht, wenn sich eines der Elektronen beliebig weit vom Kern entfernt. Außerdem ist sie über den gesamten Zweielektronenraum nicht normierbar. Diese einfache Summe zweier Einteilchenorbitale ist daher kein geeigneter Ansatz. Summen von Hartree-Produkten können dagegen durchaus sinnvolle Zweielektronenwellenfunktionen ergeben.

Um den Produktansatz zu motivieren, betrachten wir zunächst ein vereinfachtes Modell, in dem wir die gegenseitige Abstoßung der Elektronen vernachlässigen. In atomaren Einheiten lautet der Hamiltonoperator dann

$$
H_0 = h(1) + h(2), \qquad
h = -\frac12 \nabla^2 - \frac{Z}{r},
$$

wobei für Helium $Z=2$ gilt. Jeder Einteilchenoperator wirkt nur auf die Koordinaten des zugehörigen Elektrons. Deshalb erhalten wir

$$
\begin{aligned}
H_0\,\Psi
&= [h(1)+h(2)]\,\phi(\mathbf r_1)\phi(\mathbf r_2) \\
&= [h(1)\phi(\mathbf r_1)]\,\phi(\mathbf r_2)
 + \phi(\mathbf r_1)\,[h(2)\phi(\mathbf r_2)].
\end{aligned}
$$

Wählen wir $\phi$ als Eigenfunktion des Einteilchenoperators,

$$
h\,\phi = \varepsilon\,\phi,
$$

so ergibt jeder der beiden Terme den Beitrag $\varepsilon\Psi$. Damit folgt unmittelbar

$$
H_0\,\Psi = 2\varepsilon\,\Psi.
$$

Die Gesamtenergie ist also die Summe der beiden Einteilchenenergien.

Wählen wir für $\phi$ das exakte wasserstoffähnliche 1s-Orbital zur Kernladungszahl $Z$, so ist das Hartree-Produkt die **exakte räumliche Grundzustandswellenfunktion des vereinfachten Hamiltonoperators $H_0$**. Dies liefert eine natürliche Motivation für den Produktansatz. Wir sehen also, dass wir mithilfe des Produktansatzes ein Zweielektronenproblem in zwei Einteilchenprobleme zerlegen können. Dies ist die generelle Idee die wir immer wieder anwenden werden.

Beim tatsächlichen Heliumatom kommt die Elektronenabstoßung $1/r_{12}$ hinzu. Für dieses wechselwirkende System ist ein einzelnes Hartree-Produkt keine exakte Lösung, kann aber als Ansatz im Variationsverfahren verwendet werden. Warum die Produktform hier nicht exakt sein kann, zeigen wir im nächsten Abschnitt.

## Warum ein einzelnes Hartree-Produkt für Helium nicht exakt ist

Wir müssen nun zwei unterschiedliche Aussagen auseinanderhalten:

| Aussage | Form der Wellenfunktion | Bedeutung |
|:--|:--|:--|
| Separation in kollektive Koordinaten | $\chi(\mathbf R)\phi(\mathbf r)$ | Schwerpunkt und innere Bewegung werden getrennt. |
| Ein einzelnes räumliches Hartree-Produkt | $f(\mathbf r_1)g(\mathbf r_2)$ | Jedem Elektron wird eine eigene Ortsfunktion zugeordnet. |

Die Rechnung am Kernpotential zeigt, warum die erste Form die elektronische Heliumgleichung nicht wie beim freien Paar in zwei unabhängige Gleichungen zerlegt. Für die zweite Form können wir direkt beweisen, dass sie keine exakte räumliche Eigenfunktion des wechselwirkenden Heliumatoms sein kann.


### Einsetzen eines Produktansatzes

Nehmen wir an, eine exakte Eigenfunktion hätte die Form

$$
\Psi(\mathbf r_1,\mathbf r_2)=f(\mathbf r_1)g(\mathbf r_2).
$$

Wir setzen sie in $\hat H_{\mathrm{He}}\Psi=E\Psi$ ein und dividieren durch $fg$. Dies geschieht in einem offenen Bereich, in dem beide Faktoren ungleich null sind und keine Teilchen zusammenfallen. Dann folgt

$$
\underbrace{-\frac12\frac{\nabla_1^2f}{f}
-\frac{Z}{r_1}}_{A(\mathbf r_1)}
+\underbrace{-\frac12\frac{\nabla_2^2g}{g}
-\frac{Z}{r_2}}_{B(\mathbf r_2)}
+\frac{1}{r_{12}}=E.
$$

Der erste Ausdruck hängt nur von $\mathbf r_1$ ab, der zweite nur von $\mathbf r_2$. Die Gleichung würde daher verlangen, dass

$$
\frac{1}{r_{12}}=E-A(\mathbf r_1)-B(\mathbf r_2).
$$

Die Paarwechselwirkung müsste also als Summe zweier Einteilchenfunktionen darstellbar sein.


### Der Widerspruch durch gemischte Ableitungen

Eine gemischte Ableitung nach einer kartesischen Komponente $r_{1a}$ des ersten und einer Komponente $r_{2b}$ des zweiten Elektrons verschwindet für jede solche Summe:

$$
\frac{\partial^2}{\partial r_{1a}\partial r_{2b}}
\left[E-A(\mathbf r_1)-B(\mathbf r_2)\right]=0.
$$

Für den Coulomb-Term gilt dagegen mit $\mathbf r=\mathbf r_1-\mathbf r_2$:

$$
\frac{\partial^2}{\partial r_{1a}\partial r_{2b}}\frac{1}{r}
=\left(\frac{\delta_{ab}}{r^3}-\frac{3r_ar_b}{r^5}\right).
$$

Diese Matrix ist für $r>0$ nicht die Nullmatrix. Beispielsweise ist ihre Komponente entlang des Relativvektors gleich $-2/r^3$. Wir erhalten einen Widerspruch zur geforderten Produktform.

$$
\boxed{\text{Ein einzelnes räumliches Hartree-Produkt ist keine exakte Helium-Eigenfunktion.}}
$$

Dies ist eine Aussage über die räumliche Wellenfunktion. Die zusätzliche Forderung nach der richtigen Austausch- und Spinsymmetrie wird in den späteren Kapiteln behandelt. Für den Heliumgrundzustand ist eine symmetrische räumliche Funktion zusammen mit einem antisymmetrischen Spinzustand zulässig; die räumliche Coulomb-Korrelation bleibt trotzdem erforderlich. Der obige Beweis betrifft ein einzelnes Produkt und ist kein allgemeiner Beweis über beliebige Summen von Produkten oder über alle denkbaren Koordinatentransformationen.


## Energie des Heliumatoms ohne Elektron–Elektron-Abstoßung

Für den 1s-Grundzustand gilt

$$
\varepsilon = E_{1s}(Z) = -\frac{Z^2}{2},
$$

und somit ergibt sich für das Gesamtsystem ohne Elektron–Elektron-Abstoßung

$$
E = 2\,E_{1s}(Z) = 2\!\left(-\frac{Z^2}{2}\right) = -Z^2.
$$

Für das Heliumatom ($Z = 2$) folgt daher für die Gesamtennergie $E = -4$ Hartree. Der **exakte nichtrelativistische Wert bei festgehaltenem Kern** aus hochpräzisen numerischen Rechnungen beträgt

$$
E_\text{exakt} \approx -2.9037\ \text{Hartree},
$$

und aus der **Summe der beiden experimentell bestimmten Ionisationsenergien** ergibt sich für den Heliumgrundzustand eine Gesamtenergie von etwa

$$
E_\text{exp} \approx -2.9034\ \text{Hartree}.
$$

Die Differenz von rund **1.1 Hartree** zeigt deutlich, dass die einfache Näherung ohne Elektron–Elektron-Abstoßung das System **zu stark bindet**.  

Sie verdeutlicht damit den erheblichen Beitrag der Elektron–Elektron-Wechselwirkung zur realen Bindungsenergie des Heliumatoms! Wir lernen daraus, dass die Elektron-Elektron Wechselwirkung definitiv keine vernachlässigbare Größe ist und die **orbitalbasierte Produktansätze**, die für nichtwechselwirkende Elektronen exakt sind, keine gute Näherungen für die exakte Wellenfunktion eines realen wechselwirkenden Systems sind!

### Wissenswerte Ergänzung: Experimenteller Zugang zur Gesamtbindungsenergie

Wie lässt sich diese Bindungsenergie experimentell bestimmen? Einen frühen Zugang untersuchten deutsche Physiker **James Franck und Paul Knipping im Jahr 1919**. 

Sie beschleunigten Elektronen durch eine veränderliche Spannung $U$ und ließen sie mit Heliumatomen zusammenstoßen. Die kinetische Energie der einfallenden Elektronen beträgt näherungsweise $E_{\mathrm{kin}}=eU$. Gemessen wurde der Kollektorstrom als Funktion der Beschleunigungsspannung.

Rechts zeigt ein Energieniveaudiagramm den Abstand zwischen dem Heliumgrundzustand und dem vollständig ionisierten Zustand.](figures/helium_binding_experiment/helium_binding_experiment_slide.png)

*Illustration des Messprinzips. Die Apparatur ist schematisch dargestellt.*

Bereits unterhalb der Schwelle zur Entfernung beider Elektronen entsteht durch einfache Ionisation ein Strom. Franck und Knipping beobachteten eine schwache zusätzliche Änderung des Stromverlaufs, die sie der direkten Doppelionisation zuordneten. Sie bestimmten dafür eine Schwellenenergie von $(79{,}5\pm0{,}3)\,\mathrm{eV}$. 

Für den heutigen Vergleich verwenden wir die erste und die zweite Ionisationsenergie, $I_1\approx24{,}5874\,\mathrm{eV}$ und $I_2\approx54{,}4178\,\mathrm{eV}$. Wählen wir den vollständig getrennten Zustand $\mathrm{He}^{2+}+2e^-$ als Energienullpunkt, so gilt

$$
E_{\mathrm{exp}}=-(I_1+I_2)\approx-79{,}005\,\mathrm{eV}\approx-2{,}9034\,E_{\mathrm h}.
$$

Die positive Gesamtbindungsenergie ist also der Energieaufwand zur vollständigen Ionisation; die Energie des gebundenen Grundzustands hat das entgegengesetzte Vorzeichen.

## Erste Verbesserung: Energieerwartungswert des Hartree-Produktansatzes

Wir berechnen als Nächstes den **Erwartungswert des vollen Hamiltonoperators** für das Heliumatom unter Verwendung des Hartree-Produktansatzes. Das heißt, wir berechnen die Energie mit Hilfe einer genäherten Wellenfunktion, verwenden dafür aber nun den vollständigen Hamiltonoperator inklusive der Elektron–Elektron-Wechselwirkung.

Mit einem solchen Produkt als Testfunktion können wir die mittlere Elektronenabstoßung berechnen. Es fehlt aber die Anpassung der räumlichen Verteilung eines Elektrons an den jeweiligen Ort des anderen. Dies ist die hier relevante Anwendung des {ref}`zuvor eingeführten Korrelationsbegriffs <korrelation-und-statistische-unabhangigkeit>`.

Für eine gegebene (nicht notwendigerweise exakte) Wellenfunktion $\Psi$ ist der Energieerwartungswert definiert als

$$
E[\Psi] = \frac{\langle \Psi | H | \Psi \rangle}{\langle \Psi | \Psi \rangle}.
$$
 
Für das Heliumatom lautet der volle Hamiltonoperator

$$
H = h(1) + h(2) + \frac{1}{r_{12}}, 
\qquad 
h = -\tfrac12 \nabla^2 - \frac{Z}{r}.
$$

Setzen wir nun den Hartree-Produktansatz

$$
\Psi(\mathbf r_1,\mathbf r_2) = \phi(\mathbf r_1)\,\phi(\mathbf r_2),
\qquad
\phi(\mathbf r) = N\,e^{-Z r},
$$

ein, so ergibt sich für den Energieerwartungswert

$$
E[\Psi] =
\frac{
\langle \phi(1)\phi(2) |\, h(1) + h(2) + \tfrac{1}{r_{12}}\, | \phi(1)\phi(2) \rangle
}{
\langle \phi(1)\phi(2) | \phi(1)\phi(2) \rangle
}.
$$

Da $h(1)$ und $h(2)$ auf verschiedene Koordinaten wirken, kann der Ausdruck in einen **Einteilchenanteil** und einen **Wechselwirkungsanteil** zerlegt werden:

$$
E[\Psi] =
2\,\langle \phi | h | \phi \rangle
+
J,
$$

wobei das **Coulomb-Integral**

$$
J = 
\iint
|\phi(\mathbf r_1)|^2\,\frac{1}{r_{12}}\,|\phi(\mathbf r_2)|^2
\,\mathrm d^3 r_1\,\mathrm d^3 r_2
$$

den mittleren Beitrag der Elektron–Elektron-Abstoßung beschreibt.  
Damit ergibt sich der Energieerwartungswert zu

$$
E[\Psi] = 2\langle \phi | h | \phi \rangle + J.
$$

Erst durch die Berücksichtigung dieses Coulomb-Integrals $J$ liefert das Variationsprinzip eine realistische Näherung für die Gesamtenergie des Heliumatoms.



## Wie groß ist die Elektron–Elektron-Wechselwirkung? Eine erste Abschätzung.

Nachdem wir die Grundenergie des Heliumatoms **ohne** Elektron–Elektron-Abstoßung berechnet haben, wollen wir nun abschätzen, **wie groß der Beitrag des Coulomb-Integrals $J$ tatsächlich ist**.  
Wir berechnen dazu den Erwartungswert des Terms $\tfrac{1}{r_{12}}$ mit derselben Produktwellenfunktion

$$
\Psi(\mathbf r_1,\mathbf r_2) = \phi(\mathbf r_1)\,\phi(\mathbf r_2),
\qquad
\phi(\mathbf r) = N\,e^{-Z r}.
$$

Für das normierte Orbital gilt

$$
1=\int |\phi(\mathbf r)|^2\,\mathrm d^3r
=4\pi N^2\int_0^\infty r^2e^{-2Zr}\,\mathrm dr
=\frac{\pi N^2}{Z^3},
\qquad N=\sqrt{\frac{Z^3}{\pi}}.
$$

Das Coulomb-Integral lautet definitionsgemäß

$$
J = 
\big\langle \Psi \big| \frac{1}{r_{12}} \big| \Psi \big\rangle
= \iint \Psi^*(\mathbf r_1,\mathbf r_2)\,
\frac{1}{|\mathbf r_1 - \mathbf r_2|}\,
\Psi(\mathbf r_1,\mathbf r_2)\,
\mathrm d^3 r_1\,\mathrm d^3 r_2.
$$

![Koordinaten des Heliumatoms: kartesische Achsen am Kern, die Ortsvektoren der beiden Elektronen, ihr Abstand und der eingeschlossene Winkel.](figures/helium_coordinates/helium_coordinates_cartoon_v2.png)

*Die Ortsvektoren $\mathbf r_1$ und $\mathbf r_2$ haben ihren Ursprung am Kern. $r_{12}$ bezeichnet den Elektronenabstand und $\theta_{12}$ den Winkel zwischen den Ortsvektoren.*

Da es sich um ein Produkt zweier s-Orbitale handelt, ist es vorteilhaft, statt der sechs kartesischen Koordinaten $(\mathbf r_1,\mathbf r_2)$ auf **Abstandskoordinaten** zu wechseln:  
$r_1 = |\mathbf r_1|$, $r_2 = |\mathbf r_2|$ und $r_{12} = |\mathbf r_1 - \mathbf r_2|$.

Dazu führen wir zunächst Kugelkoordinaten mit dem Ursprung am Kern ein. Für das erste Elektron gilt

$$
x_1 = r_1\sin\vartheta_1\cos\varphi_1,\qquad
y_1 = r_1\sin\vartheta_1\sin\varphi_1,\qquad
z_1 = r_1\cos\vartheta_1.
$$

Die **Jacobi-Determinante** dieser Transformation liefert den Faktor

$$
\left|
\frac{\partial(x_1,y_1,z_1)}
{\partial(r_1,\vartheta_1,\varphi_1)}
\right|
= r_1^2\sin\vartheta_1,
$$

sodass das Volumenelement lautet

$$
\mathrm d^3r_1
= r_1^2\sin\vartheta_1\,
\mathrm dr_1\,\mathrm d\vartheta_1\,\mathrm d\varphi_1
= r_1^2\,\mathrm dr_1\,\mathrm d\Omega_1.
$$

Für eine festgehaltene Position des ersten Elektrons wählen wir die Polarachse für das zweite Elektron entlang von $\mathbf r_1$. Der Ursprung bleibt dabei am Kern. Diese Drehung der Achsen ändert das Volumenelement nicht. Der Polarwinkel ist nun der Winkel $\theta_{12}$ zwischen den beiden Ortsvektoren; $\varphi_{12}$ bezeichnet den Azimutwinkel um die gewählte Polarachse. Damit gilt

$$
\mathrm d^3r_2
= r_2^2\sin\theta_{12}\,
\mathrm dr_2\,\mathrm d\theta_{12}\,\mathrm d\varphi_{12}
$$

und für beide Elektronen zusammen

$$
\mathrm d^3r_1\,\mathrm d^3r_2
= r_1^2r_2^2\sin\theta_{12}\,
\mathrm dr_1\,\mathrm dr_2\,
\mathrm d\Omega_1\,\mathrm d\theta_{12}\,\mathrm d\varphi_{12}.
$$

Zwischen den drei Abständen gilt der **Kosinussatz**

$$
r_{12}^2 = r_1^2 + r_2^2 - 2r_1r_2\cos\theta_{12}.
$$

Der Integrand hängt somit nur von $r_1$, $r_2$ und $\theta_{12}$ ab. Die Orientierung von $\mathbf r_1$ und der Azimutwinkel $\varphi_{12}$ können unabhängig integriert werden:

$$
\int \mathrm d\Omega_1
= \int_0^{2\pi}\!\mathrm d\varphi_1
  \int_0^\pi\!\sin\vartheta_1\,\mathrm d\vartheta_1
= 4\pi,
\qquad
\int_0^{2\pi}\!\mathrm d\varphi_{12}=2\pi.
$$

Nach diesen Winkelintegrationen verbleibt daher das Maß

$$
8\pi^2r_1^2r_2^2\sin\theta_{12}\,
\mathrm dr_1\,\mathrm dr_2\,\mathrm d\theta_{12}.
$$

Nun ersetzen wir $\theta_{12}$ durch den Elektronenabstand $r_{12}$. Bei festgehaltenen Radien folgt durch Differenzieren des Kosinussatzes

$$
2r_{12}\,\mathrm dr_{12}
=2r_1r_2\sin\theta_{12}\,\mathrm d\theta_{12},
\qquad
\sin\theta_{12}\,\mathrm d\theta_{12}
=\frac{r_{12}}{r_1r_2}\,\mathrm dr_{12}.
$$

Die zugehörige Jacobi-Determinante für den Variablenwechsel ist im Inneren des Integrationsbereichs

$$
\left|
\frac{\partial(r_1,r_2,\theta_{12})}
{\partial(r_1,r_2,r_{12})}
\right|
=\frac{r_{12}}{r_1r_2\sin\theta_{12}}.
$$

Für die Integrationsgrenzen erhalten wir

$$
\theta_{12}=0 \;\Rightarrow\; r_{12}=|r_1-r_2|,
\qquad
\theta_{12}=\pi \;\Rightarrow\; r_{12}=r_1+r_2,
$$

woraus sich die **Dreiecksgrenzen**

$$
|r_1-r_2|\le r_{12}\le r_1+r_2
$$

ergeben. Nach der Winkelintegration und dem Variablenwechsel reduziert sich das Maß somit auf das Volumenelement

$$
\mathrm d^3r_1\,\mathrm d^3r_2
\;\longrightarrow\;
8\pi^2\,r_1r_2r_{12}\,
\mathrm dr_1\,\mathrm dr_2\,\mathrm dr_{12}.
$$

Der Pfeil erinnert daran, dass die drei Winkel für die räumliche Orientierung bereits integriert wurden.

Damit lautet das zu berechnende Integral:

$$
J =
8\pi^2N^4
\int_0^\infty\!\mathrm dr_1
\int_0^\infty\!\mathrm dr_2
\int_{|r_1-r_2|}^{r_1+r_2}\!\mathrm dr_{12}\,
r_1r_2r_{12}\,\frac{e^{-2Z(r_1+r_2)}}{r_{12}}.
$$

Der Faktor $r_{12}$ aus dem Volumenelement kürzt den Nenner des Coulomb-Terms. Die innere Integration ist deshalb besonders einfach:

$$
\int_{|r_1-r_2|}^{r_1+r_2}\!\mathrm dr_{12}
=r_1+r_2-|r_1-r_2|.
$$

Damit erhalten wir

$$
J =
8\pi^2N^4
\int_0^\infty\!\mathrm dr_1
\int_0^\infty\!\mathrm dr_2\,
r_1r_2e^{-2Z(r_1+r_2)}
\bigl(r_1+r_2-|r_1-r_2|\bigr).
$$

Da der Integrand beim Vertauschen von $r_1$ und $r_2$ unverändert bleibt, können wir uns auf den Bereich $r_1\ge r_2$ beschränken und das Ergebnis verdoppeln. In diesem Bereich gilt

$$
r_1+r_2-|r_1-r_2|=2r_2.
$$

Wir integrieren zuerst über $r_1$ von $r_2$ bis $\infty$ und anschließend über $r_2$:

$$
J =
32\pi^2N^4
\int_0^\infty\!r_2^2e^{-2Zr_2}\,\mathrm dr_2
\int_{r_2}^\infty\!r_1e^{-2Zr_1}\,\mathrm dr_1.
$$

Mit

$$
\int_{r_2}^\infty\!r_1e^{-2Zr_1}\,\mathrm dr_1
=e^{-2Zr_2}
\left(\frac{r_2}{2Z}+\frac{1}{4Z^2}\right)
$$

folgt

$$
J =
32\pi^2N^4
\left[
\frac{1}{2Z}\int_0^\infty\!r^3e^{-4Zr}\,\mathrm dr
+\frac{1}{4Z^2}\int_0^\infty\!r^2e^{-4Zr}\,\mathrm dr
\right].
$$

Die verbleibenden Integrale lassen sich mit

$$
\int_0^\infty r^n e^{-ar}\,\mathrm dr
=\frac{n!}{a^{n+1}},
\qquad a>0,\quad n=0,1,2,\ldots
$$

auswerten. Mit $N^4=Z^6/\pi^2$ ergibt sich schließlich

$$
J =
32\pi^2N^4
\left[
\frac{3!}{2Z(4Z)^4}
+\frac{2!}{4Z^2(4Z)^3}
\right]
=\frac{5\pi^2N^4}{8Z^5}
=\frac{5Z}{8}.
$$

Zum Vergleich benötigen wir zusätzlich das Normintegral

$$
\langle\Psi|\Psi\rangle =
8\pi^2N^4
\int_0^\infty\!\mathrm dr_1
\int_0^\infty\!\mathrm dr_2
\int_{|r_1-r_2|}^{r_1+r_2}\!\mathrm dr_{12}\,
r_1r_2r_{12}\,e^{-2Z(r_1+r_2)}.
$$

Hier liefert die Integration über den Elektronenabstand

$$
\int_{|r_1-r_2|}^{r_1+r_2}\!r_{12}\,\mathrm dr_{12}
=\frac12\left[(r_1+r_2)^2-(r_1-r_2)^2\right]
=2r_1r_2.
$$

Damit zerfällt das Normintegral in zwei gleiche Radialintegrale:

$$
\begin{aligned}
\langle\Psi|\Psi\rangle
&=16\pi^2N^4
\left[\int_0^\infty r^2e^{-2Zr}\,\mathrm dr\right]^2\\
&=16\pi^2N^4
\left[\frac{2!}{(2Z)^3}\right]^2
=\frac{\pi^2N^4}{Z^6}=1.
\end{aligned}
$$

Im folgenden Code verwenden wir die unnormierte Produktfunktion $e^{-Z(r_1+r_2)}$. Deshalb teilen wir das berechnete Coulomb-Matrixelement durch das Normintegral. Die Normierungskonstante kürzt sich dabei heraus.

## Zusatz: Symbolische Berechnung des Erwartungswertes 

Für **interessierte** Studierende lässt sich die Energie des Heliumatoms mit vereinfachten Hamiltonian (ohne Elektron–Elektron-Abstoßung) auch symbolisch mit sympy berechnen.  
Wir wählen als Näherung für die räumliche Wellenfunktion das Produkt zweier exponentiell abfallender 1s-Funktionen

$$
\Psi(\mathbf r_1,\mathbf r_2) = e^{-Z(r_1 + r_2)}.
$$

Da wir s-Zustände betrachten,können die Winkelintegrationen für beide Elektronen sofort durchgeführt werden:  
Das **Volumenelement** separiert zu

$$
\mathrm d^3 r = r^2 \sin\theta\, \mathrm dr\, \mathrm d\theta\, \mathrm d\phi,
\qquad \int \mathrm d\Omega = 4\pi,
$$

und für zwei Elektronen ergibt sich nach Winkelintegration das radiale Maß

$$
(4\pi)^2 \int_0^\infty\!\!\int_0^\infty r_1^2 r_2^2\, f(r_1,r_2)\,\mathrm dr_1\,\mathrm dr_2,
$$

wobei die konstanten Faktoren $(4\pi)^2$ in der Energieratio herauskürzen und daher ohne Einfluss bleiben.

Der **Laplace-Operator** reduziert sich für sphärisch symmetrische Funktionen auf seine radiale Form

$$
\nabla^2 f(r) = \frac{\mathrm d^2 f}{\mathrm d r^2} + \frac{2}{r}\frac{\mathrm d f}{\mathrm d r},
$$

sodass der nichtrelativistische Hamiltonoperator

$$
H_0 = -\tfrac12 (\nabla_1^2 + \nabla_2^2) - \frac{Z}{r_1} - \frac{Z}{r_2}
$$

im Code direkt mit radialen Ableitungen implementiert werden kann.

Das folgende **SymPy-Skript** führt dies aus, integriert über beide Elektronenradien analytisch und berechnet schließlich den Erwartungswert der Energie symbolisch:

```python
import sympy as sp
from sympy import oo
r1, r2, Z = sp.symbols('r1 r2 Z', positive=True, real=True)

def laplacian_radial(psi, r):
    return psi.diff(r, 2) + (2/r)*psi.diff(r)

def T(psi):
    return -sp.S.Half * (laplacian_radial(psi, r1) + laplacian_radial(psi, r2))

def V_ne(psi):
    return -(Z/r1 + Z/r2) * psi

def H0(psi):
    return T(psi) + V_ne(psi)

psi0 = sp.exp(-Z*(r1 + r2))  

def integrate_6d(f):
    integrand = (4*sp.pi)**2 * r1**2 * r2**2 * f
    I_r2 = sp.integrate(integrand, (r2, 0, sp.oo))
    I_r1 = sp.integrate(I_r2, (r1, 0, sp.oo))
    return sp.simplify(I_r1)

def energy(psi, H):
    num = integrate_6d(psi * H(psi))
    den = integrate_6d(psi * psi)
    return sp.simplify(num/den)

E_Z = energy(psi0, H0)            
E_Z
```

```python
E_He = sp.simplify(E_Z.subs(Z, 2))  # Helium-Ergebnis

print(f"Energie = {E_He} Hartree")
```

```python
import sympy as sp
r1, r2, r12, Z = sp.symbols('r1 r2 r12 Z', positive=True, real=True)

# Produktansatz der Wellenfunktion
psi0 = sp.exp(-Z*(r1 + r2))

# Winkelintegriertes Volumenelement in (r1, r2, r12)
dV = 8*sp.pi**2 * r1 * r2 * r12

# Normintegral S = <Psi|Psi>
S = sp.integrate(
    dV * psi0**2,
    (r12, sp.Abs(r1 - r2), r1 + r2),
    (r2, 0, sp.oo),
    (r1, 0, sp.oo)
)

# Erwartungswert <V_ee> = <Psi|1/r12|Psi>
Vee = sp.integrate(
    dV * psi0**2 * (1/r12),
    (r12, sp.Abs(r1 - r2), r1 + r2),
    (r2, 0, sp.oo),
    (r1, 0, sp.oo)
)

# Verhältnis <V_ee>/<Psi|Psi>
result = sp.simplify(Vee/S)
print(result)
```

Die symbolische Berechnung ergibt schließlich

$$
J = \frac{5Z}{8}.
$$

Für Helium ($Z=2$) beträgt das Coulomb-Integral also  $J = 1{.}25$ Hartree. Addiert man diesen Beitrag zur Energie zweier unabhängiger Elektronen ($E_0 = -4$ Hartree), erhält man für die Gesamtenergie

$$
E \approx -4 + 1{.}25 = -2{.}75\ \text{Hartree},
$$

was bereits eine vernünftige Näherung des exakten Grundzustandswerts  
$E_\text{exakt} = -2{.}9037$ Hartree darstellt. Damit wird deutlich, dass die Elektron–Elektron-Wechselwirkung einen Beitrag von etwa **1 Hartree**, also rund einem Drittel der gesamten Bindungsenergie des Heliumatoms, liefert. 
 

## Zweite Verbesserung des Ansatzes: Optimierung der effektiven Kernladung

Wie wir gerade gesehen haben, lässt sich die Energie als eine Funktion der Kernladung $Z$ wie folgt ausdrücken:

$$
E(Z) = 2\langle \phi|h|\phi\rangle + J = -Z^{2} + \frac{5}{8}Z.
$$

Dies legt nahe, die feste Kernladung $Z$ durch einen **variationellen Exponenten** $\zeta$ zu ersetzen, während der wahre Kernterm im Hamiltonoperator $-\tfrac{Z}{r}$ unverändert bleibt. Damit erhalten wir einen Energieausdruck

$$
E(\zeta;Z) = 2\Big(\tfrac{\zeta^{2}}{2} - Z\zeta\Big) + \frac{5}{8}\zeta
= \zeta^{2} - 2Z\zeta + \frac{5}{8}\zeta,
$$

den wir nun im Bezug auf die effektive Kernladung $\zeta$ optimieren können.  
Die Optimierung der effektiven Kernladung folgt aus der Bedingung

$$
\frac{dE}{d\zeta} = 2\zeta - 2Z + \frac{5}{8} = 0
\quad \Rightarrow \quad
\zeta^{\star} = Z - \frac{5}{16}.
$$

Damit ergibt sich für die optimierte Energie

$$
E^{\star}(Z) = E(\zeta^{\star};Z)
= -Z^{2} + \frac{5}{8}Z - \frac{25}{256}.
$$

Für Helium ($Z = 2$) erhält man

$$
\zeta^{\star} = 1.6875, \qquad
E^{\star} = -2.8477\ \text{Hartree},
$$

eine deutliche Verbesserung gegenüber dem unoptimierten Wert $E(Z) = -2.75\ \text{Hartree}$. Vergleicht man diesen Wert mit der exakten Grundzustandsenergie $E_\text{exakt} = -2.9037\ \text{Hartree}$, so beträgt der verbleibende Fehler nur etwa **0.056 Hartree**, also weniger als **2 %**.
Die folgende Abbildung zeigt die Energie als Funktion der effektiven Kernladung $\zeta$ bei festem $Z=2$. Die gestrichelte Linie markiert die exakte Grundzustandsenergie.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({"font.family": "DejaVu Serif", "font.size": 12,
                     "mathtext.fontset": "dejavuserif", "figure.dpi": 150})
Z_He = 2

def E(zeta):
    return zeta**2 - 2*Z_He*zeta + 5*zeta/8

zeta = np.linspace(1.1, 2.3, 300)
zeta_opt = Z_He - 5/16
E_exakt = -2.9037

fig, ax = plt.subplots(figsize=(7.5, 4.8), layout="constrained")
ax.plot(zeta, E(zeta), color="blue", lw=2.5)
ax.plot(zeta_opt, E(zeta_opt), "o", color="red", ms=8,
        label=fr"Minimum: $\zeta^\star={zeta_opt:.4f}$, $E={E(zeta_opt):.4f}\,E_\mathrm{{h}}$")
ax.plot(Z_He, E(Z_He), "o", color="blue", mfc="white", ms=8, mew=2,
        label=fr"Unoptimiert: $\zeta={Z_He}$, $E={E(Z_He):.4f}\,E_\mathrm{{h}}$")
ax.axhline(E_exakt, color="black", ls="--", lw=1.5,
           label=fr"Exakter Grundzustand: $E\approx{E_exakt:.4f}\,E_\mathrm{{h}}$")
ax.set(xlabel=r"Effektive Kernladung $\zeta$", ylabel=r"Energie $E$ (Hartree)",
       title="Helium: Variation der effektiven Kernladung",
       xlim=(zeta[0], zeta[-1]), ylim=(-2.94, -2.35))
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", color="lightgray", lw=0.8)
ax.set_axisbelow(True)
ax.legend(loc="upper center", frameon=False, fontsize=11)

from matplotlib.ticker import FormatStrFormatter

ax.yaxis.set_major_formatter(FormatStrFormatter("%.4f"))

plt.show()
```

### Explizite Abhängigkeit vom Elektronenabstand

Das [Variationsprinzip](variational_principle.ipynb) erlaubt uns, auch erweiterte Ansätze systematisch zu optimieren. In diesem Kapitel haben wir dies zunächst mit einem Orbitalprodukt durchgeführt. Die anschließende [Hylleraas-Behandlung](correlation_hylleraas.ipynb) ergänzt eine explizite Abhängigkeit vom Elektronenabstand, beispielsweise

$$
\Psi(\mathbf r_1,\mathbf r_2)
=N e^{-\zeta(r_1+r_2)}(1+\gamma r_{12}).
$$

Für $\gamma=0$ erhalten wir wieder ein Produkt; für $\gamma\ne0$ kann die Funktion die relative Anordnung der Elektronen berücksichtigen. $N$ normiert die Wellenfunktion, während $\zeta>0$ und $\gamma$ Variationsparameter sind.

<div style="background-color:#e6f0ff; border:2px solid #3366cc; border-radius:8px; padding:10px; margin-top:1em;">
<b>Geeignete kollektive Koordinaten können wechselwirkende Probleme in unabhängige Bewegungen zerlegen.</b><br>
Für Helium bleibt das innere elektronische Problem gekoppelt. Ein unabhängiges Elektronenbild ist deshalb eine Näherung; eine exakte räumliche Beschreibung erfordert eine korrelierte Zweielektronen-Wellenfunktion.

## Aufgaben zur Vertiefung

**1. Trennung trotz Wechselwirkung.** Überprüfen Sie die Zerlegung für zwei Coulomb-abstoßende Elektronen in der harmonischen Falle. Welche Masse und Frequenz hat der Schwerpunktoszillator? Warum folgt daraus weder ein harmonisches relatives Problem noch die Unabhängigkeit der beiden Elektronenorte?

**2. Warum gerade ein quadratisches äußeres Potential?** Betrachten Sie in einer Dimension zwei gleiche Massen mit derselben zweimal stetig differenzierbaren äußeren Potentialfunktion $v$ auf $\mathbb R$. Zeigen Sie

$$
\frac{\partial^2}{\partial R\partial r}
\left[v(R+r/2)+v(R-r/2)\right]
=\frac12\left[v''(R+r/2)-v''(R-r/2)\right].
$$

Folgern Sie, dass eine globale additive Zerlegung in eine Funktion von $R$ und eine Funktion von $r$ nur möglich ist, wenn $v''$ konstant ist. Welche Form hat dann $v$? Weshalb widerspricht dieses Ergebnis nicht der Separation eines freien Coulomb-Paares?


### Ergebnisse und Lösungshinweise

**Zu 1.** Der Schwerpunkt ist ein dreidimensionaler harmonischer Oszillator mit Masse $M=2$ und Frequenz $\omega$ in atomaren Einheiten. Im relativen Operator verbleibt $1/r$. Eine Produktfunktion in Schwerpunkt- und Relativkoordinaten ist im Allgemeinen kein Produkt in den Elektronenkoordinaten.

**Zu 2.** Da $R+r/2$ und $R-r/2$ unabhängig alle reellen Werte annehmen können, muss $v''$ überall denselben Wert besitzen. Zweimaliges Integrieren ergibt $v(x)=ax^2+bx+d$. Umgekehrt ist dann

$$
v(R+r/2)+v(R-r/2)=2aR^2+\frac a2r^2+2bR+2d.
$$

Diese Aussage betrifft zwei gleiche **äußere** Potentiale. Eine Paarwechselwirkung $U(x-y)=U(r)$ hängt bereits ausschließlich von der Relativkoordinate ab und darf auch nichtquadratisch sein.
