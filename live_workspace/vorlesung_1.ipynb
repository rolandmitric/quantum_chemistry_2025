```python
import sympy as sp
```

```python
beta, alpha, hbar, m = sp.symbols('beta alpha hbar m', real = True, positive = True)
x = sp.symbols('x', real = True)
```

```python
psi = sp.exp(-beta*x**2)
Norm = sp.integrate(psi * psi, (x, -sp.oo, sp.oo))
psi = psi / sp.sqrt(Norm)
psi
```

```python
sp.diff(psi,x,x,x,x,x,)
```

```python
def tpsi(psi, x, hbar, m):
    return -hbar**2/(2*m) * sp.diff(psi, x, x)
def vpsi(psi, V):
    return V * psi
def hpsi(psi, x, hbar, m, V):
    return tpsi(psi, x, hbar, m) + vpsi(psi, V)
```

```python
V = alpha * x**4
Hpsi = hpsi(psi, x, hbar, m, V)
Hpsi
```

```python
E = sp.integrate(psi * Hpsi, (x, -sp.oo, sp.oo))
E
```

```python
n = 5
psi = [x**i*sp.exp(-beta*x**2) for i in range(n)]
```

```python
def T_ij(psi_i, psi_j):
    return hbar**2 / (2 * m) * sp.integrate(sp.diff(psi_i, x) * sp.diff(psi_j, x), 
                                            (x, -sp.oo, sp.oo))

def V_ij(psi_i, psi_j, V):
    return sp.integrate(psi_i * V * psi_j, (x, -sp.oo, sp.oo))

def H_ij(psi_i, psi_j, V):
    return T_ij(psi_i, psi_j) + V_ij(psi_i, psi_j, V)

def S_ij(psi_i, psi_j):
    return sp.integrate(psi_i * psi_j, (x, -sp.oo, sp.oo))
```

```python
H = sp.zeros(n, n)
S = sp.zeros(n, n)
for i in range(n):
    for j in range(n):
        H[i, j] = H_ij(psi[i], psi[j], V)   
        S[i, j] = S_ij(psi[i], psi[j])
```

```python
beta_opt = sp.solve(sp.diff(E, beta), beta)[0]
H = H.subs(beta, beta_opt)
S = S.subs(beta, beta_opt)
H = H.subs({m: 1, hbar: 1, alpha: 1})
S = S.subs({m: 1, hbar: 1, alpha: 1})
```
